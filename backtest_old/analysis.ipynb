```python
import pandas as pd
```

```python
path ='REALISED TRADE. ETF NAME LICNETFGSC.txt'
```

```python
def read_trade_data(filename):
    """
    Reads trade data from a text file and creates a pandas DataFrame.

    Args:
        filename: Path to the text file containing trade data.

    Returns:
        pandas.DataFrame: DataFrame containing trade data.
    """

    data = []
    with open(filename, 'r') as file:
        for line in file:
            if line.startswith("REALISED TRADE"):
                # Extract data from the line
                parts = line.strip().split(',')
                
                # Parse data
                etf_name = parts[0].split(':')[1].strip()
                profit = int(parts[1].split(':')[1].strip())
                capital = int(parts[2].split(':')[1].strip())
                buy_dates_str = parts[3].split(':')[1].strip()
                buy_dates_str = buy_dates_str[1:]
                buy_dates_str = buy_dates_str[:-1]
                buy_dates_str  = buy_dates_str.split(',')[0].strip()
                # sell_date = parts[4].split(':')[1].strip()
                sell_date = line.strip().split(':')[-1]
                
                # Create a dictionary for each trade
                trade_data = {
                    'ETF NAME': etf_name,
                    'PROFIT': profit,
                    'CAPITAL': capital,
                    'BUY DATES': buy_dates_str,
                    'SELL DATE': sell_date
                }
                data.append(trade_data)

    # Create DataFrame from the list of dictionaries
    df = pd.DataFrame(data)
    # Convert buy dates string to list
    df['BUY DATES'] = pd.to_datetime(df['BUY DATES'], format='mixed')
    df['SELL DATE'] = pd.to_datetime(df['SELL DATE'], format='mixed')
    df['DAYS'] =  (df['SELL DATE'] - df['BUY DATES']).dt.days
    return df

# Example usage
filename = path  # Replace with the actual filename
df = read_trade_data(filename)

print(df)
df.to_csv('analysis.csv')
```
